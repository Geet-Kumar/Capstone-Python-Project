```python
# Read the Employee CSV file
employee = pd.read_csv("Employee.csv")

# Read the Seniority Level CSV file
seniority = pd.read_csv("Seniority_Level.csv")

# Read the Project CSV file
project = pd.read_csv("Project.csv")
```

```python
# Display the Project DataFrame
project
```

```python
# Check the number of missing values in each column
project.isnull().sum()
```

```python
# Loop through each row in the Project DataFrame
for i in range(len(project)):
    
    # Check if the Cost value in the current row is missing
    if pd.isna(project.loc[i, "Cost"]):
        
        # Calculate the average of all previous Cost values
        running_average = project.loc[:i-1, "Cost"].mean()
        
        # Replace the missing Cost value with the running average
        project.loc[i, "Cost"] = running_average
```

```python
# Display the updated Project DataFrame
project
```

```python
# Display the Employee DataFrame
employee
```

```python
# Split the Name column into First Name and Last Name
employee[["First Name", "Last Name"]] = employee["Name"].str.split(" ", expand=True)
```

```python
# Remove the original Name column
employee.drop("Name", axis=1, inplace=True)
```

```python
# Display the updated Employee DataFrame
employee
```

```python
# Merge Employee and Seniority DataFrames using the ID column
employee_seniority = pd.merge(employee, seniority, on="ID")
```

```python
# Merge the Employee-Seniority DataFrame with the Project DataFrame using the ID column
Final = pd.merge(employee_seniority, project, on="ID")
```

```python
# Display the Final DataFrame
Final
```

```python
# Create a Bonus column with 0 as the default value
Final["Bonus"] = 0
```

```python
# Give a 5% bonus to employees who have finished their projects
Final.loc[Final["Status"] == "Finished", "Bonus"] = Final["Cost"] * 0.05
```

```python
# Display the updated Final DataFrame
Final
```

```python
# Find the IDs of employees who have at least one project with status "Failed"
failed_ids = Final.loc[Final["Status"] == "Failed", "ID"].unique()

# Demote these employees by 1 level (level number increases because 1 is the highest and 4 the lowest)
Final.loc[Final["ID"].isin(failed_ids), "Designation Level"] += 1
```

```python
# Remove employee records whose Designation Level is above 4 (they lose eligibility to head projects)
Final = Final[Final["Designation Level"] <= 4].copy()
```

```python
# Display the updated Final DataFrame
Final
```

```python
# Add "Mr." to the First Name of male employees
Final.loc[Final["Gender"] == "M", "First Name"] = "Mr. " + Final["First Name"]
```

```python
# Add "Mrs." to the First Name of female employees
Final.loc[Final["Gender"] == "F", "First Name"] = "Mrs. " + Final["First Name"]
```

```python
# Remove the Gender column
Final.drop("Gender", axis=1, inplace=True)
```

```python
# Display the updated Final DataFrame
Final
```

```python
# Get the unique employees (one row per ID) with their Age
for emp_id in Final["ID"].unique():

    # Read the age of the current employee
    age = Final.loc[Final["ID"] == emp_id, "Age"].iloc[0]

    # IF the employee is older than 29 years
    if age > 29:

        # Promote by 1 level (level number decreases because 1 is the highest grade)
        Final.loc[Final["ID"] == emp_id, "Designation Level"] -= 1
```

```python
# Display the updated Final DataFrame
Final
```

```python
# Calculate the total project cost for each employee
TotalProjCost = Final.groupby(["ID", "First Name"], as_index=False)["Cost"].sum()
```

```python
# Rename the Cost column to Total Cost
TotalProjCost.rename(columns={"Cost": "Total Cost"}, inplace=True)
```

```python
# Display the Total Project Cost DataFrame
TotalProjCost
```
